```python
import matplotlib.pyplot as plt

grids = ['7x7', '12x12', '20x20']

sarl_3_agents = [79, 187, 631]
sarl_10_agents = [62, 91, 153]

jason_3_agents = [46, 40, 82]
jason_10_agents = [43, 54, 87]

plt.figure(figsize=(10, 6)) 


plt.plot(grids, sarl_3_agents, label='SARL - 3 Agents', color='#005293', linestyle='-', marker='o', linewidth=2, markersize=8)
plt.plot(grids, sarl_10_agents, label='SARL - 10 Agents', color='#005293', linestyle='--', marker='s', linewidth=2, markersize=8)

plt.plot(grids, jason_3_agents, label='Jason - 3 Agents', color='#e03a3e', linestyle='-', marker='o', linewidth=2, markersize=8)
plt.plot(grids, jason_10_agents, label='Jason - 10 Agents', color='#e03a3e', linestyle='--', marker='s', linewidth=2, markersize=8)
plt.title('Performance Scaling: SARL vs. Jason', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Grid Size (Spatial Complexity)', fontsize=12, fontweight='bold')
plt.ylabel('Execution Time (seconds)', fontsize=12, fontweight='bold')

for i, v in enumerate(sarl_3_agents):
    plt.text(i - 0.05, v + 15, str(v), color='#005293', fontweight='bold')
for i, v in enumerate(sarl_10_agents):
    plt.text(i - 0.05, v - 25, str(v), color='#005293', fontweight='bold')
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(title='Framework & Configuration', fontsize=10, title_fontsize='11', loc='upper left')i
plt.tight_layout()

plt.savefig('performance_chart.png', dpi=300)

print("saved graph 'performance_chart.png'")
plt.show()
```
